```python
articles = []
```

```python
for article in entries :
    title = article.find("{http://www.w3.org/2005/Atom}title")
    article_data = {"title" : title.text}
    articles.append(article_data)
```

```python
articles=[]
```

```python
for article in entries :
    title = article.find("{http://www.w3.org/2005/Atom}title")
    summary = article.find("{http://www.w3.org/2005/Atom}summary")
    link = article.find("{http://www.w3.org/2005/Atom}id")
    published = article.find("{http://www.w3.org/2005/Atom}published")
    article_data = {"title":title.text , "summary":summary.text , "link":link.text , "published":published.text }
    articles.append(article_data)
```

```python
for i in top_indices: 
    print("Title : " , df.iloc[i] ["title"])
    print("Summary : " , df.iloc[i] ["summary"])
    print("Link : " , df.iloc[i] ["link"])
    print("Published : " , df.iloc[i] ["published"])
    print("=============================================================")
```

```python
for i in top_indices:
    print(similarity[0,i]) # "0" means row of query and "i" means articles from 0 to 99
```

```python
def recommend_articles (query):
    query_vector = vectorizer.transform([query])
    similarity = cosine_similarity(query_vector , tfidf_matrix)
    top_indexes = similarity[0].argsort()[: : -1] [0:5]
    for i in top_indices: 
        print("Title : " , df.iloc[i] ["title"])
        print("Summary : " , df.iloc[i] ["summary"])
        print("Link : " , df.iloc[i] ["link"])
        print("Published : " , df.iloc[i] ["published"])
        print("=============================================================")   

    
```

```python
recommend_articles("natural language processing")
```

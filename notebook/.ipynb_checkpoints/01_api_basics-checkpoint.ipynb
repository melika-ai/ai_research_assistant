```python
url = "https://export.arxiv.org/api/query?search_query=nlp&start=0&max_results=5"
```

```python
response = requests.get(url)
```

```python
response.status_code
```

```python
response.text
```

```python
print(response.text[:1000])
```

```python
import xml.etree.ElementTree as ET
```

```python
root = ET.fromstring(response.text)
```

```python
root 
```

```python
len(root)
```

```python
entries = root.findall("{http://www.w3.org/2005/Atom}entry")
```

```python
len(entries)
```

```python
first_article = entries[0]
```

```python
first_article
```

```python
for item in first_article:
    print(item.tag)
```

```python
title = first_article.find("{http://www.w3.org/2005/Atom}title")
```

```python
title
```

```python
title_text = title.text
```

```python
title_text
```

```python
for article in entries:
    title = article.find("{http://www.w3.org/2005/Atom}title")
    print(title.text)
```

```python
titles = []
for article in entries:
    title = article.find("{http://www.w3.org/2005/Atom}title")
    titles.append(title.text)
    
```

```python
titles
```

```python
articles = []
```

```python
for article in entries :
    title = article.find("{http://www.w3.org/2005/Atom}title")
    article_data = {"title" : title.text}
    articles.append(article_data)
```
